# Análisis visual A1/A2 - base homologada

Este notebook ejecuta el generador de visualizaciones para revisar la normalización de datos hacia la base homologada, con cortes por país, fuente, clase, uso A1.8, score y vacíos de información.

Las salidas quedan en `resultados/`: un dashboard HTML, figuras SVG y tablas CSV resumidas.

```python
from pathlib import Path
import runpy

candidates = [
    Path.cwd() / "generar_visualizaciones_homologacion.py",
    Path.cwd() / "docs" / "actividad_2" / "A2_llave_class_uso" / "analisis_visual_homologacion_a1_a2" / "generar_visualizaciones_homologacion.py",
]

SCRIPT = next((path for path in candidates if path.exists()), None)
if SCRIPT is None:
    raise FileNotFoundError("No se encontró generar_visualizaciones_homologacion.py")

runpy.run_path(str(SCRIPT), run_name="__main__")
RESULTS = SCRIPT.parent / "resultados"
RESULTS
```

## Dashboard

El dashboard es autocontenido y también puede abrirse directamente desde `resultados/dashboard_homologacion_a1_a2.html`.

```python
from IPython.display import HTML, display

dashboard = RESULTS / "dashboard_homologacion_a1_a2.html"
display(HTML(dashboard.read_text(encoding="utf-8")))
```
